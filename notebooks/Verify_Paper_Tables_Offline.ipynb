# IEEE TCSS — Verify paper tables without social-media data

This notebook checks numerical values already exported by the frozen experiments. It **does not** retrain models, re-verify source trees, or make new generalization claims. Run it from the checked-out repository.

```python
from pathlib import Path
import csv, json
P=Path.cwd()
if not (P/'results').exists() and (P.parent/'results').exists(): P=P.parent
assert (P/'results/twitter_covid19/external_summary_metrics.csv').exists(), 'Open notebook from repository root or notebooks/'
print('Repository:',P)
```

```python
with (P/'results/twitter_covid19/external_summary_metrics.csv').open(newline='') as f: r=list(csv.DictReader(f))
r={x['model']:x for x in r}
assert abs(float(r['C_shared']['macro_f1'])-0.474390595245624)<1e-10
assert abs(float(r['D_shared']['macro_f1'])-0.4604586101813459)<1e-10
assert abs(float(r['D_shared']['mcc'])-(-0.07019515600066104))<1e-10
print('External shared-threshold Macro-F1: C=',r['C_shared']['macro_f1'],' D=',r['D_shared']['macro_f1'])
```

```python
with (P/'results/ontology216_faithfulness/reports/rule_faithfulness_summary.csv').open(newline='') as f: rules=list(csv.DictReader(f))
assert len(rules)==6 and all(int(x['n_case_rule_pairs'])==216 for x in rules)
print('Verified 216 cases per rule, 1296 total interventions')
for x in sorted(rules,key=lambda y:-float(y['mean_abs_delta_p'])):print(x['rule'],round(float(x['mean_abs_delta_p']),5),round(float(x['fraction_flipped'])*216))
```

```python
with (P/'results/ontology216_faithfulness/reports/historical_prediction_reconstruction_audit.csv').open(newline='') as f: groups=list(csv.DictReader(f))
assert len(groups)==8 and all(x['exact_reference_match_at_1e-4']=='True' for x in groups)
print('Eight rebuilt-model groups agree with saved reference probabilities at tolerance 1e-4.')
```
